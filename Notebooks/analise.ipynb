### Quais são as plataformas mais vendidas ? 


```python
import sqlite3
import pandas as pd
import sys, os
sys.path.append(os.path.abspath('../src'))
import database_manager as dbm

conn = dbm.criar_conexao('vendas_games.db')
dbm.carregar_csv_to_sql('../Data/vgsales.csv', 'vendas', conn)
query = """
SELECT
    Platform,
    SUM(Global_Sales) AS Total_Vendas
FROM vendas
GROUP BY Platform
ORDER BY Total_Vendas DESC
LIMIT 10
"""
df_pergunta = pd.read_sql_query(query, conn)
print(df_pergunta)
```

### 2- Qual genêro domina as vendas ? 

```python
import os, sys 
sys.path.append(os.path.abspath('../src'))
import database_manager as dbm
import pandas as pd

conn = dbm.criar_conexao('vendas_games.db')

query = """
SELECT 
    Genre, 
    SUM(Global_Sales) AS Total_Vendas
    FROM vendas
    GROUP BY Genre
    ORDER BY Total_Vendas DESC
    LIMIT 10
"""

df_pergunta = pd.read_sql_query(query, conn)
print(df_pergunta)
```

### 3 - Como as vendas variam ao longo dos anos 

```python
import os, sys 
sys.path.append(os.path.abspath('../src'))
import database_manager as dbm
import pandas as pd

conn = dbm.criar_conexao('vendas_games.db')

query = """
SELECT 
    Year,
    SUM(Global_Sales) || ' Milhões'  AS Total_Vendas
FROM vendas
WHERE Year IS NOT NULL
GROUP BY Year
ORDER BY Year DESC
LIMIT 10
"""
df_pergunta = pd.read_sql_query(query, conn)    
print(df_pergunta)
```

### 4 - Quem são os publisher com mais participação no mercado

```python
import os, sys 
sys.path.append(os.path.abspath('../src'))
import database_manager as dbm
import pandas as pd

conn = dbm.criar_conexao('vendas_games.db')

query = """
SELECT
    Publisher,
    SUM(Global_Sales)  AS Total_Vendas
    FROM vendas
    GROUP BY Publisher
    ORDER BY Total_Vendas DESC
    LIMIT 10
"""
df_pergunta = pd.read_sql_query(query, conn)
print(df_pergunta)
```
